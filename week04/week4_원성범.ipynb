```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

wine = pd.read_csv('https://bit.ly/wine_csv_data')

data = wine[['alcohol', 'sugar', 'pH']].to_numpy()
target = wine['class'].to_numpy()

train_input, test_input, train_target, test_target = train_test_split(data, target, test_size=0.2, random_state=42)
```

```python
from sklearn.model_selection import RandomizedSearchCV # 트리를 효율적으로 훈련 ( 파인 튜닝 ) 시키는 놈 
from sklearn.ensemble import RandomForestClassifier # 트리 여러개를 합친 것

params = {'min_impurity_decrease': np.arange(0.0001, 0.001, 0.00025),
          'max_depth': range(5, 20, 3),
          'min_samples_split': range(2, 100, 10)
          }
rf = RandomForestClassifier(n_jobs=-1, random_state=42)

gs = RandomizedSearchCV(rf, params, n_jobs=-1) # 제일 앞 인자가 훈련시킬 모델이다.
gs.fit(train_input, train_target)

print(gs.best_params_) # 최적의 하이퍼파라미터 조합
print(gs.best_score_) # 최적 조합의 점수
```

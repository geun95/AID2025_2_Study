1. Momentum
    - 개념: 이전 단계의 이동 방향(기울기)을 기억해서, 현재 단계의 기울기 업데이트에 추가로 더해주는 방법이다.
    - 이전 문제: 경사 하강법은 local minimum에 빠져버리는 문제와 굴곡진 손실함수에서 진동(oscillation)이 발생해 느리게 수렴하는 문제가 있다.
    - 해결: 관성 개념을 적용해서 2개의 문제를 해결했다. 관성을 통해 local minimum에 빠졌을 때 벗어나서 global minimum의 위치까지 나아갈 수 있으며, 진동 방향은 서로 상쇄되고 진행 방향이 관성에 의해 강화되어 더 효율적으로 수렴한다.
2. Adagrad
    - 개념: 각 파라미터 별로 학습률을 다르게 조절하는 방법이다. Adaptive Gradient.
    - 이전 문제: 경사 하강법은 모든 Feature마다 동일한 학습률을 사용한다. 그러나 파라미터마다 중요도, 크기 등이 다 다르기 때문에 비효율적이다.
    - 해결: 각 파라미터의 과거 기울기 정보를 누적해 학습률을 자동으로 조절한다. 누적이 커지면 학습률이 작아진다.
3. RMSProp
    - 개념: 과거 기울기의 지수 이동 평균을 사용해 각 파라미터의 학습률을 조절하는 방법이다.
    - 이전 문제: Adagrad에서는 단순히 과거 기울기 제곱합을 누적해서 학습률이 점점 줄어들다가 멈춰버리는 문제가 있다.
    - 해결: 기울기의 지수 이동 평균을 사용해서 최근 기울기는 많이 반영하고 먼 과거의 기울기는 조금만 반영한다.
4. Adam
    - 개념: Momentum과 RMSProp을 결합한 방법이다.
    - 이전 문제: 기존 방법들은 방향 정보와 학습률 조절을 따로 처리했다.
    - 해결: 기울기의 1차 모멘트(기울기 평균)와 2차 모멘트(기울기 제곱 평균)를 동시에 추정해 파라미터별로 학습률을 자동으로 조절한다.
5. AdamW
    - 개념: 가중치 감소를 통해 Adam의 과대적합을 방지하는 방법이다.
    - 이전 문제: Adam에서 L2 정규화를 적용할 때 모멘트 기반 스케일링으로 인해 정규화가 제대로 되지 않아 과대적합이  발생한다.
    - 해결: 가중치 감쇠 항이랑 Adam의 업데이트 식을 분리해서 정규화 효과가 제대로 반영되게 했다. 더 안정적이고 일반화된 학습이 가능하다.

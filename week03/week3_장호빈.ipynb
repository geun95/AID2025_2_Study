모멘텀(Momentum)은 경사하강법이 골짜기 모양의 손실곡면에서 좌우로 심하게 흔들리며 느리게 내려가는 문제를 줄이기 위해, 이전 단계의 이동방향을 속도처럼 누적해 관성으로 활용하는 방법이다. 기울기의 지수이동평균을 따라가므로 수직 방향 진동이 완화되고, 유효한 하강 방향에는 가속이 붙는다(보통 모멘텀 계수 β≈0.9).

AdaGrad는 차원마다 학습률을 다르게 두어 희소 특성에서 자주 등장하는 특징은 학습률을 줄이고, 드물게 등장하는 특징은 학습률을 키우는 방식으로 스스로 스케일을 맞춘다. 이를 위해 과거 제곱기울기를 누적해 분모에 넣어 스텝 크기를 조정하는데, 누적이 계속 커지면 학습률이 지나치게 작아져 후반에 학습이 멈추는 경향이 있다.

RMSProp은 AdaGrad의 “누적이 너무 커져서 학습률이 고갈된다”는 약점을 해결하려고, 제곱기울기의 단순 누적 대신 지수이동평균(EMA)을 사용한다. 최근 기울기에 더 큰 가중을 줘 분모가 과도하게 커지는 것을 막아 비정상(non-stationary)한 손실 지형에서도 안정적으로 학습률을 유지한다(감쇠계수 ρ≈0.9).

Adam은 모멘텀의 1차 모멘트(평균 기울기)와 RMSProp의 2차 모멘트(제곱기울기 평균)를 결합하고, 초기 단계 편향을 보정해 빠른 수렴과 적응형 스케일 조정을 동시에 노린다. 기본 하이퍼파라미터(β1≈0.9, β2≈0.999, ε 작은 값)만으로도 폭넓은 과제에서 잘 작동해 사실상 기본값처럼 쓰인다. 다만 전통적인 “L2 정규화”를 그대로 섞으면 적응스케일과 뒤엉켜 의도한 가중감쇠 효과와 달라질 수 있다.

AdamW는 이 점을 바로잡기 위해 Adam의 파라미터 업데이트와 가중감쇠(weight decay)를 분리(decouple) 했다. 즉, 정규화를 “기울기 스케일 조정과 독립적인 별도 항”으로 적용해 정규화의 의미를 보존하고 일반화 성능을 안정적으로 얻는다. 이 때문에 최근 비전·언어 대규모 모델 학습에서 AdamW가 표준 선택지로 자리 잡았다.

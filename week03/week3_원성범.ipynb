## 옵티마이저란?
* 확률적 경사 하강법에서 모델이 산의 정상에서 밑으로 내려갈 때, 걸음 걸이를 더 똑똑하게 조절하는 법에 대한 것이다.

### 1. Momentum
* 개념 : 굴러가는 공은 멈추기 힘들다, 관성을 이용한다.
* 기존 문제 : 기본 SGD는 경사가 조금만 바뀌어도 방향이 이리저리 흔들린다.
    * 그래서 속도가 느리고,
    * 음푹 파인 곳을 잘 못빠져 나온다.
* 아이디어 : 방금 내려온 방향의 속도를 기억해서 다음 걸음에서 추진력을 받는다. -> 자잘한 흔들림은 무시하게 된다. 

### 2. Adagrad
* 개념 : 처음 걷는 길은 성큼성큼가고, 익숙한 길에서는 조심히 가는 것
* 기존 문제 : 모든 변수에 같은 보폭을 적용하는 것이 비효율적이다.
    * 경사가 가파르면 ( 많이 변했던 변수 ) 보폭을 줄이고,
    * 완만하면 ( 적게 변한 변수 ) 보폭을 늘리자.
* 아이디어 :
    * '지금까지' 많이 변했던 변수는 잘 알고 있으므로, 보폭을 줄여서 조심한다.
    * '지금까지' 많이 안 변했던 변수는, 잘 모르는 길이니까 일단 성큼 가보는 식이다.
* 단점 : 훈련을 너무 오래했을 때 -> 기록이 계속 쌓이면 모든 변수의 보폭이 0에 수렴해버림.

### 3. RMSProp
* 개념 : Adargrad의 학습 중단을 해결함. 과거는 적당히 잊고, '최근'에 집중
* 기존 문제 : 모든 과거를 기억한 Adargrad의 학습 중단
* 아이디어 : '지금까지'의 모든 변화를 똑같이 더하기 X, 최근의 변화량에만 가중치를 둬서 평균내기.

### 4. Adam
* 개념 : Momentum + RMSProp 
* 기존 문제 : 빠른 가속도 + 효율적인 보폭 조절을 합침
* 아이디어 : 1번 주머니에서는 관성을 기억해서 내려갈 방향을 부드럽게, 2번 주머니에서는 '최근'변화량을 따로 기억해서 변수마다 보폭을 조절한다.

### 5. AdamW
* 개념 : Adam을 더 안정적으로 개선. 정답만 외우는게 아니라, 일반 문제도 더 잘 풀도록 한다.
* 기존 문제 : Adam이 좋긴 한데, 종종 과대적합으로 인해 실제 테스트 데이터에서 성능이 안 나오기도 한다.
* 아이디어 : 가중치 감쇠 ( adamW의 W가 weight decay임 )를 이용한다. 모델의 변수들이 너무 복잡하게 커지는 것을 막아준다. 암기만 하는 것을 방지. 일반적인 성능을 높여줌 ( 일반화 )

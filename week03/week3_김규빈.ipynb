Momentu, Adagrad, RMSProp, Adam, AdamW

## Momentu  

**개념**  
- 관성을 더해서 평균적으로 내리막인 방향으로 가속, 불필요한 진동 x


**해결하려는 문제**
- 미니배치 경사 하강법등에서는 노이즈 때문에 경사가 들쭉날쭉함 -> 지그재그로 진동하게됨 -> global minimum 까지 가는데에 오래걸림
- 깊은 local minimam에 빠져버리면 나오지 못하고 이를 global minimum으로 오해해 버릴수도
- 한쪽은 급경사, 한쪽은 완만한 구조에서 특히 이런 문제가 심해짐


**해결 아이디어**
- 수식에 가속도*가중치 값을 더하여 관성을 부여함(가중치는 보통 1보다 작은 0.9사용)
- 관성에 의해 local minimum에 들어가도 빠져나올 수 있고, 방향성이 조금 더 일정해지므로 지그재그로 진동하는 경향도 줄어듬


## Adagrad

**개념** 
- 많이 학습된 파라미터는 조금 학습되도록, 조금 학습된 파라미터는 많이 학습되도록 파라미터 마다 학습률의 차등을 둠

**해결하려는 문제**
- 각 파라미터 마다 중요도, 크기 등이 다르기 때문에 일괄적으로 동일한 학습률을 적용하는건 비효율적일 수 있음
- 가끔씩만 나타나는 특징이 충분히 학습되지 않고 무시될 수 있음

**해결 아이디어**
- 과거 기울기 제곱의 누적합을 분모에 넣어서 많이 학습된 파라미터 일수록 학습률을 줄임
- 따라서 큰 변화가 있었던 파라미터의 학습률은 매우 작아지고, 작은 변화가 있었던 파리미터의 학습률은 조금 작아짐

## RMSprop

**개념**
- Adagrad에서 과거의 기울기 변화는 조금 반영, 최근의 기울기 변화는 많이 반영

**해결하려는 문제**
- Adagrad의 경우에는 학습이 많이 진행되면 분모가 커지면서 학습률이 점점 0으로 수렴해버린다는 문제점 존재

**해결 아이디어**
- 기울기를 같은 비율로 누적하지 않고 지수이동평균을 활용
- 과거의 기울기 변화는 조금만 반영하고 최근의 기울기 변화는 많이 반영함으로써 학습률이 0이되는 Adagrad의 문제점을 해결


## Adam 

**개념**
- Momentum과 RMSprop의 장점을 섞음 -> 학습의 방향과 크기를 모두 최적화

**해결하려는 문제**
- Momentum에서 학습의 방향을 최적화하고 RMSporp에서 학습의 크기를 최적화함 -> 두가지를 모두 잡아보자

**해결 아이디어**
- 관성과 학습률을 모두 계산해서 학습에 반영
- 초반에 관성과 학습률이 0으로 편향되는 경향이 있기에 편향 보정을 거쳐서 해결

## AdamW

**개념**
- Adam에서 규제(Weight decay)를 더욱 정확하게 적용

**해결하려는 문제**
- Adam은 L2 정규화가 학습률을 계산하기 위한 스케일 보정에 섞여버리면서 규제 효과가 왜곡될 수 있음

**해결 방안**
- 파라미터 업데이트와 규제를 분리해서 적용 -> L2 정규화항이 스케일 보정과 섞이지 않음
- 규제가 모든 파라미터에 균일하게 적용되어 일반화 성능 증가

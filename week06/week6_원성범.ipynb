### 1. 가중치 초기화 (Weight Initialization)

- **역할**: 학습의 시작점 설정. 기울기 소실(Vanishing)이나 폭주(Exploding)를 방지하여 학습이 가능하게 함.
- **사용법**: 활성화 함수에 맞춰 선택.
    - **ReLU 계열**: `He Initialization` (현재 표준)
    - **Sigmoid/Tanh**: `Xavier (Glorot) Initialization`

---

### 2. 배치 정규화 (Batch Normalization)

- **역할**: 각 레이어를 통과할 때마다 데이터 분포가 흐트러지는 것을 잡아주어 학습을 안정화하고 속도를 높임.
- **사용법**: 일반적으로 활성화 함수 **앞**에 배치.
    - 순서: `Conv/Linear` → `Batch Norm` → `Activation (ReLU)`
    - **주의**: 학습(Train) 모드와 추론(Eval) 모드를 구분해서 사용해야 함.

---

### 3. 학습률 스케줄링 (LR Scheduling)

- **역할**: 학습 진행 상황에 맞춰 학습률(Learning Rate)을 동적으로 조절하여 최적점(Global Minima)에 잘 도달하게 함.
- **사용법**:
    - **Warm-up**: 초반에 학습률을 천천히 올림 (안정적 출발).
    - **Decay**: 학습이 진행될수록 학습률을 줄임 (정교한 탐색).
    - 종류: `StepLR`, `CosineAnnealing` 등.

---

### 4. 데이터 증강 (Data Augmentation)

- **역할**: 데이터를 인위적으로 변형하여 데이터 양을 늘리고, 모델의 과적합(Overfitting)을 방지(일반화 성능 향상).
- **사용법**:
    - 기법: `Crop`(자르기), `Flip`(뒤집기), `Rotation`(회전), `Mixup`(이미지 섞기) 등.
    - **주의**: **훈련(Train) 데이터셋에만 적용**하고, 검증/테스트 셋에는 원본을 유지해야 함.

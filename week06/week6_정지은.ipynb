1. 배치 정규화
    
    : 배치 단위 별로 데이터를 평균과 분산을 이용해 정규화하는 것이다.
    
    - 역할: layer가 깊어지면서 입력 분포가 바뀌는 현상을 완화해 학습을 안정적이게 해준다.
    - 사용법: layer 바로 뒤에 Batch Normalization Layer를 연결해야 한다. 그 뒤에 활성화 함수를 둔다. 학습 단계랑 추론 단계에서 평균과 분산을 다르게 적용해야 한다. 학습 단계에서는 현재 배치의 통계값으로 정규화하고, 추론 단계에서는 학습 중 누적된 running mean과 running variance를 사용한다.
2. 학습률 스케줄링
    
    : 학습률을 고정하지 않고 훈련이 진행될수록 변화시키는 기술이다.
    
    - 역할: 최적점 주변에서 학습률을 점점 줄여서 안정적으로 수렴하도록 한다.
    - 사용법: scheduler 객체를 만들어서 optimizer에 연결해서 자동으로 학습률이 조정되게 한다.
3. 데이터 증강
    
    : 기존의 데이터를 변형해서 새로운 데이터를 만들어내는 기술이다.
    
    - 역할: 데이터가 부족할 때 사용하거나 모델이 과대 적합 되는 것을 방지하기 위해 사용한다.
    - 사용법: 훈련 데이터에만 적용해야 한다. 딥러닝 프레임워크에서 제공하는 클래스를 사용해 회전, 반전, 이동, 스케일링, 잘라내기, 색 변화 등 원하는 대로 데이터를 변형하면 된다. 보통 학습 중에 매 batch마다 실시간으로 랜덤 변형을 적용한다.
4. 가중치 초기화
    
    : 학습을 시작할 때 가중치를 어떤 값으로 설정할지 결정하는 기술이다.
    
    - 역할: 적절한 값으로 초기화해서 gradient 흐름이 안정된다.
    - 사용법: layer를 생성할 때 kernel_initializer 매개변수로 initializer를 지정한다. custom layer를 생성할 경우에는 init 함수를 호출해서 초기화한다.

딥러닝 훈련 기술을 조사하고 각각의 역할과 사용법을 정리하기.   
(배치 정규화, 학습률 스케줄링, 데이터 증강, 가중치 초기화 등..)

## 배치정규화

**역할**  

각 레이어마다 scale이 커지거나 mean이 편향되는 등 출력분포가 흐트러지면 학습이 불안정하게 됨.  
이를 해결하고자 레이어의 입력 값을 평균 0, 분산 1에 가깝게 정규화

**사용법**

- 신경망의 각 레이어에서 활성화 함수를 적용하기 전 또는 후에 배치 정규화 레이어를 삽입 
- 훈련때 : 미니 배치의 평균과 분산을 계산해 정규화 
- 추론때 : 학습단계에서 업데이트된 평균과 분산으로 정규화



## 학습률 스케일링

**역할**

학습 초기에는 큰 학습률로 빠르게 학습 -> 이후 학습률을 줄여서 정밀하게 학습  
이를 통해 학습속도 및 성능을 개선

**사용법**

- optimizer와 scheduler를 생성해 학습루프에서 이를 호출하여 학습 속도를 조절한다
- 다양한 scheduling 방법이 존재
- StepLR: 일정 에포크마다 학습률을 동일하게 감소시킴
- ExponentialLR: 에포크마다 지속적으로 학습률을 지수배 감소시킴
- CosineAnnealingLR: 코사인 함수 형태로 부드럽게 학습률을 감소시켰다가 다시 증가시키는 주기를 반복
- ReduceLROnPlateau: 모델의 성능이 일정 시간 동안 개선되지 않았을 때 학습류를 자동으로 감소




## 데이터 증강

**역할**

원본 데이터를 데이터 변환연산을 통해 인위적으로 확대
이를 통해 데이터 다양성을 증가시켜 일반화 성능을 향상시키고, 과적합을 방지

**사용법**

- 이미지 데이터: 회전, 좌우반전, 자르기, 밝기, 채도 조절 등
- 텍스트 데이터: 동의어 대체, 단어순서 변경, 무작위 삽입/삭제 등


## 가중치 초기화

**역할**

모델 학습을 시작할 때 가중치를 어떤 값으로 설정할지 정함  
학습 초기에 기울기 소실, 기울기 폭발 등의 문제를 방지


**사용법**

- Xavier 초기화: 입력 데이터의 분산과 출력 데이터의 분산이 각 층을 통과하며 일정하게 유지되도록 가중치의 초기 분산을 설정
- he 초기화: 활성화 함수로 ReLU를 사용할때 주로 사용, Xavier보다 더 넓은 분산을 가짐

### DNN과 CNN 모델의 성능 및 훈련 속도 비교와 분석

* 성능 및 속도 비교 성능은 CNN이 DNN보다 훨씬 뛰어남. 특히 이미지 데이터의 경우 DNN은 1차원 배열로 펴서 입력받느라 공간 정보를 잃어버리지만, CNN은 이미지의 형상을 유지하며 학습하기 때문임. 훈련 속도 면에서도 CNN이 더 효율적임. 단순 연산 속도보다는 유의미한 성능을 내기 위해 필요한 가중치(파라미터) 수가 CNN이 훨씬 적어서 수렴이 빠름.

* 차이가 발생하는 이유 분석 첫째, 지역적 연결성 때문임. DNN은 픽셀 하나가 다음 층의 모든 뉴런과 연결돼서 낭비가 심하고 과적합되기 쉬움. 반면 CNN은 필터가 이미지의 특정 영역만 훑으며 특징을 잡음. 둘째, 가중치 공유 때문임. DNN은 위치마다 다른 가중치를 쓰지만, CNN은 하나의 필터를 이미지 전체에 공유해서 씀. 덕분에 학습할 파라미터 수가 획기적으로 줄어들고 학습 효율이 높음.
